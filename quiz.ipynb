### Вопрос 1
1. Зачем нужна кнопка History и какие функции появляются при нажатии правой кнопкой мыши на коммит?
Ответ:
History показывает историю коммитов — сохранённых изменений проекта.

- Copy SHA — скопировать идентификатор коммита.
- View on GitHub — открыть коммит на GitHub.

### Вопрос 6

**Как создать и отправить коммит?**

**Ответ:** Сохранить файлы Ctrl+S → добавить нужные изменения в Staged Changes (кнопка +) → написать сообщение коммита → Commit → Push/Sync Changes. Commit записывает изменения локально, Push отправляет их на GitHub.

### Вопрос 7

**Как убедиться, что коммит отправлен на GitHub?**

**Ответ:** Открыть свой репозиторий на сайте GitHub, выбрать правильную ветку, открыть Commits и найти свой коммит по названию или SHA. Если коммит виден в удалённой ветке, Push выполнен.

### Вопрос 8

**Какая команда показывает, что код не прошёл проверки?**

**Ответ:** pre-commit run --all-files — запускает все настроенные pre-commit-проверки для файлов репозитория. При сбое конкретная проверка получает статус Failed и в выводе появляется сообщение об ошибке.

---

### Вопрос 8.1

**Какие линтеры и средства проверки используются?**

**Ответ:** В .pre-commit-config.yaml учебного репозитория настроены Pylint (статический анализ), Flake8 (ошибки и стиль), Mypy (типы), Pydocstyle (docstring), Codespell (опечатки). Есть также Black (форматирование), Isort (импорты), Pyupgrade (обновление синтаксиса), Docformatter (формат docstring), Jupytext (синхронизация ipynb/py) и обёртки nbQA для проверок ноутбуков. Не все перечисленные инструменты являются линтерами в строгом смысле.

### Вопрос 22

**Что такое stash?**

**Ответ:** git stash временно прячет незакоммиченные изменения из рабочей папки, чтобы можно было переключить ветку или принять новые изменения, а потом вернуть свою незавершённую работу.

---

### Вопрос 22.1

**Как сохранить stash через Git и VS Code?**

**Ответ:** Команда: `git stash push -u -m "незавершённая работа"` (опция `-u` также сохраняет новые, неотслеживаемые файлы; без неё они остаются в рабочей папке). В VS Code: Ctrl+Shift+P → Git: Stash (либо меню Source Control → Stash, если доступно).

---

### Вопрос 22.2

**Как восстановить stash?**

**Ответ:** git stash list → git stash apply stash@{0} (возвращает, но не удаляет stash) или git stash pop (возвращает и удаляет запись при успешном применении). В GitHub Desktop — Stashed Changes → Restore.

---

### Вопрос 22.3

**Чем stash отличается от commit? Когда что применять?**

**Ответ:** Commit создаёт постоянную точку в истории локальной ветки; stash — временно хранит незавершённые изменения вне истории ветки. Для готового логического результата нужен commit, для временного переключения задачи — stash.

### Вопрос 22.4

**Как посмотреть список stash?**

**Ответ:** Команда: git stash list.

---

### Вопрос 22.5

**Как удалить отдельный stash или все stash?**

**Ответ:** Отдельный: git stash drop stash@{0}. Все: git stash clear. Эти команды удаляют сохранённые изменения из stash, поэтому использовать их нужно осторожно.

---

### Вопрос 22.6

**Примеры, когда помогает stash**

**Ответ:** Нужно срочно перейти на другую ветку; нужно сделать git pull при незакоммиченных изменениях; хочешь ненадолго проверить другую версию проекта, не потеряв текущую работу.

---

### Вопрос 23

**Где увидеть конфликты файлов?**

**Ответ:** В VS Code в Source Control (раздел Merge Changes), внутри файла с конфликтными блоками, а также в GitHub Desktop при слиянии; команда git status показывает unmerged paths.

---

### Вопрос 23.1

**Когда возникает конфликт?**

**Ответ:** Когда Git не может автоматически совместить изменения двух версий — например, в одной и той же строке файла есть разные правки при merge, rebase или восстановлении stash.

### Вопрос 24

**Как решить конфликт файлов?**

**Ответ:** Открыть конфликтный файл, сравнить изменения, выбрать подходящий вариант или вручную объединить оба, удалить конфликтные маркеры, проверить итог, git add, затем завершить merge-коммитом или кнопкой Complete Merge.

---

### Вопрос 25

**Какие изменения синие, а какие зелёные в показанном редакторе конфликтов?**

**Ответ:** В примере из учебных ответов: зелёным обозначены пришедшие изменения (incoming), синим — текущие локальные (current). Цвета могут отличаться в других версиях или темах редактора; ориентироваться лучше по надписям Incoming и Current.

---

**Ответ:** Да. Stash освобождает рабочую папку; затем можно сделать git pull. Нужно учитывать, что при возвращении stash возможен конфликт.

**Какая команда преобразует ipynb в py?**

### Вопрос 2

**Что такое пакетный менеджер? pip или conda, что лучше для DS?**

**Ответ:** Пакетный менеджер устанавливает и обновляет зависимости. pip ставит Python-пакеты; conda управляет ещё и окружениями и некоторыми не-Python-библиотеками. Для Data Science conda удобна при сложных научных зависимостях, но pip в виртуальном окружении тоже полностью подходит. В этой работе использован pip.

---

### Вопрос 3

**Почему .py иногда удобнее, чем .ipynb?**

**Ответ:** Файл .py содержит обычный текст Python-кода, поэтому его легче сравнивать в Git, ревьюить и использовать в автоматизированных проверках. .ipynb удобен для экспериментов, графиков и пошагового обучения — он не хуже во всех ситуациях.

---

### Вопрос 4

**Что такое PEP 8?**

**Ответ:** Официальное руководство по стилю Python-кода: отступы, пробелы, длина строк, именование и др. https://peps.python.org/pep-0008/

---

### Вопрос 4.1

**Линтеры проверяют соблюдение PEP 8?**

**Ответ:** Да, некоторые — например Flake8 и Pylint — проверяют часть стилевых правил PEP 8. Black автоматически форматирует код, но не является линтером.

### Вопрос 4.2

**Какая нотация для названий переменных?**

**Ответ:** По PEP 8 для обычных переменных и функций используется snake_case: например, total_price. Имена классов обычно используют CapWords (PascalCase).

---

### Вопрос 4.3

**Может ли переменная называться одной буквой или _?**

**Ответ:** Да. Например, x = 5; символ _ — допустимое имя и часто обозначает значение, которое намеренно не используется.

---

### Вопрос 4.4

**Для чего используют подчёркивание _?**

**Ответ:** В snake_case: total_price; для неиспользуемого значения: for _ in range(3); ведущий _ в имени — условный признак внутреннего использования; в числах 1_000_000 — разделитель разрядов.

---

### Вопрос 4.5

**Разрешает ли PEP 8 переменную из одной буквы?**

**Ответ:** Да, прямого запрета нет. Но имена l (маленькая латинская L), O (большая O) и I (большая i) в качестве однобуквенных переменных нежелательны, потому что легко путаются с цифрами 1 и 0. В остальных случаях важна понятность.

---

### Вопрос 4

**Какие способы переноса строк есть в задании?**

**Ответ:** В задании перечислены шесть способов, показанных ниже: обратный слэш, тройные кавычки, список с join, круглые скобки, f-строки в скобках и объединение через +.

---

### Вопрос 4.1

**Перенос с обратным слэшем**

**Ответ:** Символ \ позволяет продолжить выражение на следующей физической строке, но этот приём сравнительно хрупкий.

```python
string_continued = (
    "This is a long string that we want to " "split across multiple lines."
)
print(string_continued)
```

### Вопрос 4.2

**Многострочный текст в тройных кавычках**

**Ответ:** Тройные кавычки сохраняют внутренние переводы строк, поэтому результат отличается от простого разбиения длинной строки.

```python
multi_line_string = """This is a string that spans
multiple lines. It keeps the line breaks."""
print(multi_line_string)
```

---

### Вопрос 4.3

**Объединение списка строк через join**

**Ответ:** Метод join объединяет отдельные фрагменты; разделитель \n вставляет перевод строки между ними.

```python
strings = ["First line.", "Second line.", "Third line."]
result = "\n".join(strings)
print(result)
```

---

### Вопрос 4.4

**Продолжение строки с круглыми скобками**

**Ответ:** В скобках соседние строковые литералы автоматически объединяются; это удобный способ для длинного текста.

```python
long_string = (
    "This is a very long string that I would like to " "continue on the next line."
)
print(long_string)
```

---

### Вопрос 4.5

**Форматированные f-строки в скобках**

**Ответ:** В f-строке выражения в фигурных скобках подставляют значения переменных.

```python
letter_a = 5
letter_b = 6
message = (
    f"When {letter_a} is multiplied by {letter_b}, "
    f"the result is {letter_a * letter_b}."
)
print(message)
```

---

### Вопрос 4.6

**Сложение строк знаком +**

**Ответ:** Оператор + соединяет две строки; используется, когда объединяются уже готовые строковые значения.

```python
string_part1 = "This is the first part, "
string_part2 = "and this is the second part."
full_string = string_part1 + string_part2
print(full_string)
```

**Ответ:** Git — распределённая система контроля версий, работающая локально. Коммиты хранят историю, можно посмотреть разницу и вернуться к выбранному состоянию. GitHub — удалённый сервис для совместной работы.

**Как вставить изображение в Markdown?**

**Ответ:** Синтаксис: ![Описание](путь/картинка.png). Картинку нужно сохранить в проекте и правильно указать путь или URL.

### Вопрос 9

**Ответ:** В строке состояния редактора есть Ln и Col (строка и позиция курсора). Для ориентира длины можно включить editor.rulers, например на 79-й колонке.

### Вопрос 10

**Как изменить тип ячейки с Python на Markdown?**

**Ответ:** Выделить ячейку, нажать на указатель типа ячейки и выбрать Markdown. Также можно использовать меню ячейки Change Cell to Markdown.

### Вопрос 11

**Как запустить все ячейки Jupyter сразу?**

**Ответ:** В верхней панели ноутбука нажать Run All. Для выполнения кодовых ячеек потребуется выбрать Python Kernel.

### Вопрос 12

**Как поменять размер картинки? Нужен ли HTML?**

**Ответ:** Markdown сам по себе обычно не задаёт размер. Можно использовать HTML в Markdown: <img src="image.png" width="300">. Для простого изменения размера HTML знать полностью не нужно, достаточно этого тега.

### Вопрос 13

**Какой хоткей запускает ячейку и переходит к следующей?**

**Ответ:** Shift+Enter. Для Markdown он отображает разметку, а для кодовой ячейки выполняет код и переходит ниже.

### Вопрос 14

**Как включить номера строк в ячейках Notebook?**

**Ответ:** В меню ноутбука/ячейки выбрать Cell Line Numbers → On. В VS Code можно найти настройку Notebook: Line Numbers и установить on; клавиши зависят от среды.

### Вопрос 15

**Что такое Go To и как перейти к ошибке?**

**Ответ:** Команда навигации к строке, файлу или ячейке. Через Ctrl+G можно перейти к строке в обычном файле; в Notebook использовать Go to Cell или переход из Problems/сообщения об ошибке.

### Вопрос 16

**Как очистить результаты ранее запущенных ячеек?**

**Ответ:** В верхнем меню ноутбука нажать Clear All Outputs (или для конкретной ячейки — Clear Cell Outputs). Это удаляет сохранённые выводы, но не сам код.

### Вопрос 20

**Какая ошибка была в трассировке: DICT или LIST?**

**Ответ:** В доступном примере учебного решения упоминается NameError: name 'dict' is not defined. Однако имя встроенного типа dict обычно существует; возможно, в видео была переменная DICT с другим регистром. Точное написание без просмотра кадра подтвердить нельзя.

### Вопрос 22

**Что такое валидация?**

**Ответ:** Валидация — проверка, удовлетворяют ли данные, документ или результат определённым требованиям. Например, проверка допустимых типов и обязательных полей запроса.

### Вопрос 23

**Что такое трассировка ошибки?**

**Ответ:** Traceback — цепочка вызовов и строк программы, показывающая, где возникло исключение и через какие функции выполнение туда пришло.

### Вопрос 24

**Что значит «отвалился интерпретатор»?**

**Ответ:** Обычно это значит, что процесс Python или ядро Jupyter остановились, зависли или разорвалась связь VS Code с ними. Нужно посмотреть логи, выбрать правильный Kernel и при необходимости перезапустить.
