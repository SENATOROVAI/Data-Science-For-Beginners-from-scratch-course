1. Что такое CPython и чем он отличается от Python?
- CPython - это наиболее распространенная реализация Python с компилятором,
  написанная на С и компилируется С-компилятором.
- CPython используется для выполнения большинства Python-программ.
- Это не ЯП это реализация Python на языке программирования C.
- Реализация исполняет байт-код, созданный интерпретатором.
- Python описывает синтаксис, а CPython интерпретирует код по этим правилам.

6. Почему Python считается быстрым, хотя он медленнее С?
- Чистый Python медленнее С или Rust: каждая операция работает с
  полноценным объектом. Зато код пишется быстрее, а тяжёлые вычисления
  выполняются в библиотеках на С/C++ (NumPy, Pandas, OpenCV, TensorFlow
  и др.).

10. (опционально) Как работает интерпретатор CPython при выполнении кода?
Подготовительный этап(один раз):
10.1. Исходный код СPython, написанный на С, компилируется компилятором
      (Clang) и линкуется в исполняемый файл python.

Во время выполнения программы:
10.2 Интерпретатор CPython читает исходный .py
10.3 Выполняет лексический анализ, синтаксический анализ (парсинг) и строит
     AST (абстрактное синтаксическое дерево)
10.4 Компилирует AST в байткод Python.
10.5 Python Virtual Machine последовательно интерпретирует инструкции
     байткода.
10.6 Для выполнения каждой инструкции используется соответствующие функции
     СPython, написанные на языке С, которые взаимодействуют с памятью,ОС и
     процессором.

Введение в PyPy

14. Чем PyPy отличается от CPython?
- PyPy гораздо быстрее CPython за счет JIT-компиляции

15. Почему PyPy не может использоваться для всех проектов на Python?
- может не работать с библиотеками и расширениями, написанными специально на
  CPython на C.

17. Как установить PyPy после скачивания?
- распаковать скачанный архив с pypy.org в удобную папку; запускается
  исполняемым файлом pypy3 из папки bin

18. Как запустить файл с помощью PyPy?
- копируем путь интерпретатора в cmd + пробел + путь полный до файла с
  названием файла
19. Почему PyPy выполняет код быстрее, чем CPython?
- быстрее при запуске чистого кода за счет JIT-компиляции, но некоторые
  библиотеки и расширения, написанные с использованием CPython С API, могут
  работать медленнее, быть несовместимыми или требовать спец слоя
  совместимости (cpyext)
