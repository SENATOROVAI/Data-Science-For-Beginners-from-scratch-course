# GitHub

1.1. Что такое GitHub?
- веб-платформа для хранения Git-репозиториев и совместной разработки ПО. Она
  представляет инструменты для работы с кодом, Issues, PR, GitHub Actions,
  Projects, GH Docs
1.2. Как GitHub связан с Git?
- системой контроля версий, при помощи команд git можно управлять удаленными
  репозиториями в GH: получать изменения из удаленного репо и отправлять в
  него локальные коммиты.
1.3. Чем отличается fork репозитория от его клонирования (clone)?
- fork - копия чужого репо, созданная УЗ пользователя GH и сохраняющая связь
  с исходным репо. Clone - создание локальной копии любого удаленного репо на
  компьютере разработчика.
1.4. Зачем нужны и как работают pull requests?
- PR - это запрос на включение изменений из одной ветки в другую. Он
  используется для проверки кода, обсуждения изменений и их слияния с
  основной веткой проекта. Разраб создает отдельную ветку на GH и создаем PR.
  Владелец или участники проекта могут проверить код, оставить комментарии,
  запросить доработку, принять изменения или закрыть PR без слияния.

1.5. GitHub использует ваш почтовый адрес для привязки ваших Git коммитов к
     вашей учётной записи?
- Да. GitHub связывает Git-коммит с учётной записью, если адрес электронной
  почты, указанный в коммите, добавлен и подтверждён в этой учётной записи.
  Также можно использовать специальный GitHub-адрес вида noreply.

# Создание запроса на слияние
4 Можно ли открыть пул реквест, если вы ничего не вносили в FORK?
- нет. Pull Request можно открыть только при наличии изменений (коммитов) в
  вашем fork или ветке по сравнению с веткой исходного репозитория. Без
  изменений создать Pull Request нельзя.

4.1 Что нужно сделать чтобы открыть пул реквест? (Рисунок 90)
①Клонируем нашу копию
②Создаём тематическую ветку
③Вносим свои изменения
④Проверяем изменения
⑤Фиксируем изменения в тематической ветку
⑥Отправляем новую ветку в нашу копию на GitHub
Теперь, если мы зайдём на страничку нашей копии на GitHub, мы увидим, что
GitHub заметил наши изменения и предлагает открыть запрос на слияние с
помощью большой зелёной кнопки.
Также можно зайти на страницу «Branches», по адресу
https://github.com/<user>/<project>/branches, найти интересующую ветку и
открыть запрос оттуда

4.2 Что нужно сделать Если ваш Форк устарел?
- подтянуть актуальные изменения из исходного репо (upstream) и отправить
  обновленную ветку в свой fork (origin) Sync fork - Update branch

4.3 Что нужно сделать если в пул реквесте имеются конфликты слияния (Рисунок
    96)
- Если в PR возникли конфликты слияния, необходимо разрешить  их: обновить
  свою ветку актуальными изменениями из целевой ветки, вручную устранить
  конфликты в файлах, проверить рез-т, сделать новый коммит и отправить
  изменения git push. После этого PR обновится автоматически.

# Особенные файлы

9 Что такое и зачем нужен файл README
- это лицевая часть проекта с его описанием, зависимостями, требованиям,
  примеры использования, инструкцией развертывания и запуску, информация о
  лицензиях и авторах. Он помогает пользователям и разработчикам понять
  назначение проекта, способы его установки, запуска и использования.

9.1 Что такое и зачем нужен файл CONTRIBUTING (Рисунок 122)
- Если в вашем репозитории будет файл CONTRIBUTING с
любым расширением, то GitHub будет показывать ссылку на него при создании
любого
запроса на слияние.
Вы можете указать конкретные вещи, которые вы хотите или не
хотите видеть в новых запросах на слияние. Таким образом люди могут
ознакомится с
руководством, перед тем как создавать новый запрос на слияние.

10.2 Что такое файл .gitignore?
- файл, который содержит список файлов и каталогов, которые Git должен
  игнорировать и не добавлять в репо, такие как .venv, .env, __pycache__,
  большие датасеты. Он используется, чтобы не хранить временные файлы,
  зависимости, конфиденциальные и большие данные. Он действует только на
  неотслеживаемые файлы, которые не попали в индекс.
.gitignore бывает кастомным и шаблонным под каждый язык программирования
