# Квиз 1. Алгоритм сдачи домашних заданий

## Основные вопросы

**1. Зачем нужна кнопка History и какие функции появляются при нажатии правой кнопкой мыши на коммит?**

Вкладка **History** в GitHub Desktop показывает историю коммитов текущей ветки: кто, когда и что изменил. Если выбрать коммит, справа видно список изменённых файлов и сами изменения (diff): зелёным добавленные строки, красным удалённые.


**1.1. Где взять ссылку на коммит? Куда её отправлять?**

Правый клик по коммиту в History -> View on GitHub. Откроется страница коммита в браузере, ссылку копируем из адресной строки. Она имеет вид https://github.com/<user>/<repo>/commit/<SHA>.


**2. Что такое файл лога?**

Файл лога - это рабочий журнал. В нём записывается, что сделано на уроке: что изучено, какие задания выполнены, какие ошибки возникли и как они решены, какие остались вопросы. Это позволяет проверяющему и самому студенту видеть прогресс.

**2.1. Когда нужно его пушить?**

В идеале после внесения каждого измениния конце каждого урока (рабочего дня), после того как в него внесены записи о проделанной работе. Коммит с логом отправляется на GitHub вместе с выполненными заданиями.

**3. Что такое интерпретатор?**

Интерпретатор программа, которая читает код на Python и выполняет его построчно (без предварительной компиляции всего кода в машинный). Для Python стандартный интерпретатор CPython.

**4. Где можно выбрать интерпретатор?**

- В VS Code для .py файлов: в правом нижнем углу строки состояния (там написана версия Python) или через Ctrl+Shift+P -> Python: Select Interpreter.
- Для ноутбуков .ipynb : кнопка Select Kernel в правом верхнем углу ноутбука.

**5. Что такое модуль?**

Модуль это файл с кодом Python (.py), который можно импортировать в другой код командой import. В модуле лежат функции, классы и переменные. Несколько модулей, собранных в папку, образуют пакет.

**6. Как создать и отправить коммит?**

В GitHub Desktop:
1. На вкладке Changes отметить нужные файлы.
2. Внизу слева написать сообщение коммита (Summary), например docs: add git quiz.
3. Нажать Commit to <ветка>.
4. Нажать Push origin вверху.

Через терминал:

git add,
git commit -m "docs: add git quiz",
git push


**7. Как убедиться, что коммит отправлен и находится на GitHub?**

- В GitHub Desktop кнопка вверху показывает Fetch origin без стрелки с числом (если есть стрелка с числом, коммит ещё не отправлен).
- git status пишет `Your branch is up to date with 'origin/<ветка>.
- На сайте GitHub в репозитории, в нужной ветке, коммит виден в списке Commits.

**8. Какая команда показывает, что код не прошёл проверки на ошибки?**


pre-commit run --all-files
Она запускает все проверки из .pre-commit-config.yaml. Напротив каждой проверки пишется Passed, Failed или Skipped. Если есть хотя бы один Failed, код не прошёл проверку. Те же проверки запускаются автоматически при git commit.

**8.1. Перечислите линтеры, которые используются для проверки кода, и кратко охарактеризуйте каждый.**

- Ruff - быстрый линтер и форматер. Заменяет flake8, pep8-naming, pydocstyle, isort, pyupgrade и black: проверяет стиль PEP 8, имена, docstring, порядок импортов, современный синтаксис, форматирует код.
- black - форматер: приводит код к единому стилю автоматически.
- isort -  сортирует и группирует импорты.
- flake8 -  проверяет стиль PEP 8 и ошибки (неиспользуемые переменные и т.п.); в курсе с плагинами, которые запрещают однобуквенные имена переменных (VNE001) и требуют понятных имён функций.
- pylint - глубокий анализ кода: ошибки, плохие практики, дублирование, сложность.
- mypy -  проверка типов (аннотаций типов).
- pydocstyle - проверка оформления docstring (описаний модулей, функций, классов).
- pyupgrade - переписывает код на современный синтаксис Python.
- docformatter - форматирует docstring.
- codespell - ищет опечатки в английских словах.
- nbQA - позволяет запускать эти линтеры на Jupyter-ноутбуках.
- jupytext - создаёт .py -копию каждого ноутбука.

**9. Как узнать, какой именно линтер не прошёл проверку?**

В выводе pre-commit run --all-files строка с названием линтера помечена 'Failed'.

**10. Видит ли линтер Pylint markdown?**

Нет. Pylint проверяет  только код в code-ячейках. Markdown-ячейки для него невидимы.

**11. Могут ли номер ячейки в терминале и номер ячейки в VS Code отличаться? В каком случае?**

Да, могут. Линтеры через nbQA нумеруют ячейки по-своему (например, учитывают только code-ячейки), а VS Code считает все ячейки подряд, включая markdown. Поэтому если в ноутбуке есть markdown-ячейки, номера отличаются.


**12. Где посмотреть номер ячейки в VS Code?**

В строке состояния внизу справа, когда ячейка выбрана: Cell 5 of 20. Также ячейки и их порядок видны в панели Outline, а слева от code-ячейки отображается номер выполнения [5].

**13. В каком формате .ipynb отправляется на GitHub? При чём здесь JSON?**

Файл .ipynb - это текстовый файл в формате JSON. Внутри: список ячеек ( cells ), у каждой тип (code/markdown), исходный текст (source), результаты выполнения (outputs) и метаданные. На GitHub уходит именно этот JSON. Поэтому diff ноутбуков трудно читать, и поэтому линтерам нужна прослойка nbQA, а для проверки рядом создаётся.py-копия через jupytext.

**14. Где посмотреть, в какой ячейке ошибка?**

- В выводе линтера в терминале:  `file.ipynb:cell_3:5:1 означает ячейку 3, строку 5, столбец 1.
- Если ошибка при выполнении - в трассировке под ячейкой (Cell In[3], line 5), а после Run All кнопка Go To переходит к упавшей ячейке.
- В панели Problems (Ctrl+Shift+M).

**15. Как запустить терминал?**

В VS Code: меню View + Terminal или сочетание  Ctrl+ . Новый терминал Ctrl+Shift+.

**16. Что такое линтер?**

Линтер - программа для статического анализа кода: она проверяет код без его запуска и находит ошибки, опечатки, нарушения стиля , подозрительные конструкции и плохие практики. Линтеры помогают держать код единообразным и читаемым.

**17. На какой сайт нужно вставлять код ошибки, если она связана с Pylint?**

В документацию Pylint, раздел сообщений: https://pylint.readthedocs.io/en/stable/user_guide/messages/messages_overview.html. Там для каждого кода (например, C0114 missing-module-docstring) есть описание, пример неправильного и правильного кода.

**18. Секция pydocstyle в большинстве случаев закрывается автоматически после исправления ошибок в каком линтере?**

После исправления ошибок flake8: в нём через плагин flake8-docstrings проверяются те же правила docstring (коды D...), что и в pydocstyle. Эти правила объединены в Ruff (группа D), поэтому при исправлении ошибок Ruff ошибки pydocstyle исчезают тоже.

**19. Что такое описание модуля? Должно ли оно отражать содержимое модуля?**

 Это краткий поясняющий текст или блок документации в начале файла с кодом, который объясняет назначение, основные функции и правила использования этого модуля. Да, оно обязательно должно отражать содержимое модуля, описывая его суть без погружения в мелкие детали реализации.


**20. С какой git-команды начинается утро программиста?**

С git pull - забрать свежие изменения с GitHub, чтобы работать с актуальной версией и избежать конфликтов. В GitHub Desktop Fetch origin, затем Pull origin.

**21. После внесения изменений в файлы кнопка «Open in Visual Studio Code» в GitHub Desktop пропадает. Как ещё можно открыть VS Code из GitHub Desktop?**

Нажать на файл правой кнопкой мыши -> Open in Visual Studio Code или сочетанием клавиш Ctrl+Shift+A.

## Stash

**22. Что такое stash? Объясните концепцию в общих чертах.**

Stash временное хранилище для незакоммиченных изменений. Команда убирает изменения из рабочей папки (файлы возвращаются к состоянию последнего коммита) и откладывает их в отдельный стек. Потом их можно вернуть обратно в ту же или другую ветку. Это удобно, когда нужно срочно переключиться на другую задачу, а коммитить незаконченную работу не хочется.

**22.1. Как сохранить изменения в stash?**

- git-команда (подсказка):
- кнопка в VS Code:

- git-команда: стэш это буфер куда мы временно прячем файлы, чтобы принять pull git stash или с подписью git stash push -m "описание".
- кнопка в VS Code: Source Control -> ... (More Actions) -> Stash -> Stash.

**22.2. Как восстановить stash (подсказка)?**

- git-команда (подсказка):

- git stash pop - вернуть последние изменения и удалить запись из stash.
- git stash apply - вернуть, но оставить запись в stash.
- git stash apply stash@{n} - вернуть конкретную запись.

**22.3. Чем stash отличается от коммита? Когда лучше сохранить изменения в stash, а когда — закоммитить?**

- Коммит - постоянная точка в истории ветки: с сообщением, хранится в репозитории и отправляется на GitHub.
- Stash - временная «полка» только на вашем компьютере: не попадает в историю, не пушится, его легко забыть или потерять.

Stash - когда работа не закончена и нужно ненадолго переключиться (сменить ветку, сделать pull). 
Коммит - когда изменения логически завершены и их нужно сохранить надолго и отправить.

**22.4. Как просмотреть список сохранённых stash?**

- git-команда (подсказка):

git stash list. Посмотреть содержимое записи: git stash show -p stash@{0}.

**22.5. Как удалить stash? Приведите команды для удаления отдельной записи и всех сразу.**

- git-команда (подсказка):

- Одна запись: git stash drop stash@{0} (без аргумента удаляется последняя).
- Все записи сразу: git stash clear.

**22.6. Приведите практические примеры использования stash — краткие сценарии, где он помогает.**

1. Начали работу, а нужно забрать изменения с GitHub: git stash -> git pull -> git stash pop.
2. Срочно нужно исправить ошибку в другой ветке: спрятали незаконченную работу, исправили, вернулись и достали изменения.
3. Хотим проверить, работал ли код до наших правок: спрятали правки, запустили, вернули.

## Конфликты

**23. Где увидеть, что в файлах есть конфликт?**

- В VS Code: в Source Control появляется раздел Merge Changes, файлы помечены буквой !/ C; в самом файле подсвечены блоки между маркерами <<<<<<<, =======, >>>>>>>.
- В GitHub Desktop: окно с сообщением о конфликтах и списком файлов.
- В терминале: git status показывает Unmerged paths и both modified.

**23.1. Когда он возникает?**

Когда одни и те же строки одного файла изменены по-разному в двух местах (например, локально и на GitHub, или в двух ветках), и Git не может сам решить, какую версию оставить. Случается при git pull, git merge, git rebase, git stash pop.

**24. Как разрешить конфликт в файлах?**

1. Открыть файл с конфликтом.
2. Для каждого блока выбрать: Accept Current Change (оставить локальное), Accept Incoming Change (оставить пришедшее), Accept Both Changes(оставить оба) - или отредактировать вручную, удалив маркеры <<<<<<<, =======, >>>>>>>.
3. Сохранить файл и добавить его в индекс (+ в Source Control или git add <файл>).
4. Сделать коммит.

**25. Выберите правильное утверждение:**

а) зелёное - то, что пришло с GitHub, синее - локальные изменения;
б) синее - то, что пришло с GitHub, зелёное - локальные изменения.

**Правильный ответ б)** - синее (Incoming Change) пришло с GitHub, зелёное (Current Change) - локальные изменения.

**26. Если мы работаем в одном файле, можно ли принять pull после того, как спрятали свои изменения в stash?**

Да. После git stash рабочая папка чистая, поэтому git pull пройдёт без помех.

**26.1. Что может произойти, если восстановить stash после принятия pull?**

Если пришедшие с GitHub изменения затронули те же строки, что и спрятанные, при git stash pop возникнет конфликт. Его нужно разрешить как обычный конфликт. Запись при этом остаётся в stash, после разрешения её удаляют вручную git stash drop. Если строки разные, изменения объединятся.

**27. Сколько способов разрешения конфликтов показано в видео? Укажите число и перечислите способы.**

Два способа:
1. Кнопки прямо в файле: Accept Current Changes/ Accept Incoming Changes / Accept Both Changes.
2. Resolve in Merge Editor и Complete Merge.

**28. Что делает кнопка Complete Merge?**

Завершает разрешение конфликта в Merge Editor: сохраняет итоговую версию файла и отмечает конфликт как решённый
(добавляет файл в индекс, как git add). После этого остаётся сделать коммит.

## Дополнительные вопросы

**1. Какая команда конвертирует файл из .ipynb в .py? (подсказка)**


jupyter nbconvert --to script file.ipynb

или через jupytext (так делает pre-commit в курсе):

**2. Что такое пакетный менеджер? Каким вы пользуетесь — conda или pip? Какой лучше использовать в data science?**

Пакетный менеджер - программа, которая устанавливает, обновляет и удаляет библиотеки и следит за их зависимостями.

- pip - стандартный менеджер Python, ставит пакеты из PyPI.
- conda - менеджер пакетов и окружений из Anaconda/Miniconda; умеет ставить и не-Python зависимости (C-библиотеки, CUDA).

Я пользуюсь conda. 
В data science часто выбирают conda, потому что она проще ставит тяжёлые научные библиотеки.

**3. Почему расширение .py лучше, чем .ipynb?**

- .py - обычный текст: diff в Git читается легко, конфликты разрешать проще;
- линтеры, форматеры и тесты работают с .py напрямую;
- код можно импортировать как модуль и запускать из терминала;
- в .py нет выводов ячеек и метаданных, файл меньше и чище.

.ipynb удобен для исследований и визуализации, но для production-кода и code review лучше .py.

**4. Что такое PEP 8? (подсказка)**

PEP 8 - официальное руководство по стилю кода на Python (https://peps.python.org/pep-0008/): отступы в 4 пробела, длина строк, пробелы вокруг операторов, именование, порядок импортов, пустые строки и т.д.

**4.1. Проверяют ли линтеры соблюдение PEP 8?**

Да. Например, pycodestyle/flake8 и Ruff (правила E, W, N), pylint; а black и ruff format автоматически форматируют код в стиле, совместимом с PEP 8.

**4.2. Какая нотация используется для именования переменных? (ответ на стр. 85–95: https://t.me/c/1937296927/1/16676)**

snake_case - слова в нижнем регистре через подчёркивание,
camelCase - слова записываются в нижнем регистре, но второе и следующие слова начинаются с заглавной,
PascalCase - первое слово в ней тоже начинается с заглавной.

**4.3. Может ли имя переменной состоять из одного символа, например из подчёркивания _?**

Да, синтаксически это допустимое имя: `_ = 5` работает. Но однобуквенные имена запрещены линтером (flake8-variables-names, правило VNE001), поэтому лучше давать понятные имена.

**4.4. Зачем и где используется подчёркивание _?**

- Переменная-«заглушка», значение которой не нужно: for _ in range(3).
- В интерактивном интерпретаторе _ хранит результат последнего выражения.
- _init_, _str_ - специальные (dunder) методы.
- class_, type_ - чтобы не конфликтовать с ключевыми словами и встроенными именами.
- Приватные атрибуты классов всегда начинают с символа подчеркивания ( _ ).


**4.5. Допустимо ли по PEP 8 имя переменной из одной буквы? (ответ на стр. 85–95: https://t.me/c/1937296927/1/16676)**

PEP 8 прямо не запрещает однобуквенные имена, но требует никогда не использовать `l`, `O` и `I` — их легко спутать с цифрами 1 и 0. Однобуквенные имена допустимы разве что для коротких счётчиков, но хорошей практикой считаются осмысленные имена. В конфигурации линтеров курса однобуквенные имена переменных запрещены.

**4. Какие способы переноса строк показаны в видео?**

Разбить комментарий на несколько решеток, заключить в круглые скобки и перенести, разделить одну длинную переменную на несколько, а затем concatenate

**4.1. Обратный слэш (\\):**

```python
string_continued = "This is a long string that we want to " \
                   "split across multiple lines."
print(string_continued)
```

\ в конце строки говорит интерпретатору, что инструкция продолжается на следующей строке; две строки-литерала рядом склеиваются в одну.

**4.2. Тройные кавычки (''' или """):**

```python
multi_line_string = """This is a string that spans
multiple lines. You can write freely
and it will keep the line breaks."""
print(multi_line_string)
```

Строка в тройных кавычках может занимать несколько строк, и переносы строк сохраняются внутри самой строки (в ней будут символы \n). Подходит для многострочного текста и docstring, но не для «разбиения» одной длинной строки без переносов.

**4.3. Список строк, объединённый через join:**

```python
strings = [
    "This is the first line.",
    "This is the second line.",
    "This is the third line.",
]
result = "\n".join(strings)  # объединяем через перенос строки '\n'
print(result)
```

Строки хранятся в списке, а метод str.join склеивает их через разделитель - здесь \n, поэтому каждая выводится с новой строки. Удобно, когда строк много или они формируются программно.

**4.4. Круглые скобки для продолжения строки:**

```python
long_string = (
    "This is a very long string that I would like to "
    "continue on the next line."
)
print(long_string)
```

Внутри круглых скобок Python позволяет переносить строки без обратного слеша, а соседние строковые литералы автоматически склеиваются в одну строку (без символа переноса). Способ, рекомендованный PEP 8.

**4.5. f-строки в круглых скобках:**

```python
letter_a = 5
letter_b = 6
product_ab = letter_a * letter_b

message = (
    f"when {letter_a} is multiplied by {letter_b}, "
    f"the result is {product_ab}"
)
print(message)
```

То же, что 4.4, но с f-строками: в каждую часть подставляются значения переменных в {}. Каждая часть, где есть подстановка, должна иметь свой префикс f.

**4.6. Конкатенация строк через +:**

```python
string_part1 = "This is the first part, "
string_part2 = "and this is the second part."
full_string = string_part1 + string_part2
print(full_string)
```

Оператор + склеивает строки (в том числе хранящиеся в переменных). Подходит для небольшого числа строк.

**7. Что такое Git? Он локальный? Можно ли в нём посмотреть историю изменений файлов и вернуться к любому коммиту?**

Git - система контроля версий: она хранит историю изменений файлов в виде коммитов.

Да, Git локальный: весь репозиторий с полной историей хранится у вас на компьютере (в папке .git), а GitHub лишь удалённая копия для обмена и хранения.

Да, можно: история git log, git switch --detach <SHA> (посмотреть), git restore --source <SHA> <файл> (вернуть файл), git revert <SHA> (отменить коммит), git reset --hard <SHA> (откатить ветку).

**22. Что такое валидация?**

Валидация - проверка того, что данные или код соответствуют заданным правилам и формату: например, что JSON корректен (check-json в pre-commit), что в поле формы введён e-mail, что код проходит линтеры. **23. Что такое трассировка (traceback) ошибки?**

Traceback - отчёт Python об ошибке: цепочка вызовов, которая привела к ней, с указанием файлов/ячеек и строк, а в последней строке тип ошибки и сообщение (например, KeyError: 'name').
