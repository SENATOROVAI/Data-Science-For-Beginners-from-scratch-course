## 1. GitHub

**1.1. Что такое GitHub?**

GitHub — крупнейший хостинг Git-репозиториев и платформа для совместной разработки. Кроме хранения кода в нём есть issues, pull requests, ревью кода, уведомления, вики, GitHub Actions для CI/CD и управление доступом к проектам.

**1.2. Как GitHub связан с Git?**

Git — распределённая система контроля версий, она работает локально на компьютере. GitHub хранит Git-репозитории на сервере и служит для них удалённым репозиторием (remote): изменения отправляются туда через `git push` и забираются через `git fetch` / `git pull`. Сверху GitHub добавляет веб-интерфейс и инструменты для совместной работы. Без Git GitHub не работает, а Git без GitHub — работает.

**1.3. Чем fork репозитория отличается от его клонирования (clone)?**

- **Fork** — копия чужого репозитория на сервере GitHub в моём аккаунте. В форк у меня есть права на push, и он сохраняет связь с исходным проектом: из него можно открыть pull request в оригинал.
- **Clone** — локальная копия репозитория (любого: своего, чужого или форка) на моём компьютере со всей историей.

Обычно делают оба шага: сначала fork на GitHub, потом clone форка к себе.

**1.4. Зачем нужны pull request и как они работают?**

Pull request — предложение влить изменения из одной ветки (часто из ветки форка) в другую ветку проекта. Он нужен, чтобы вносить вклад без прав на запись и чтобы изменения проходили обсуждение, ревью и автоматические проверки до попадания в основную ветку.

Как работает: я создаю ветку, делаю коммиты, отправляю ветку на GitHub и открываю PR. Мейнтейнеры комментируют код построчно, CI запускает проверки. Новые коммиты в эту ветку автоматически добавляются в PR. В конце владелец делает merge или закрывает PR.

**1.5. Использует ли GitHub ваш адрес электронной почты для привязки Git-коммитов к вашей учётной записи?**

Да. GitHub сопоставляет email автора в коммите (`git config user.email`) с адресами, добавленными в аккаунт (Settings → Emails). При совпадении коммит показывается с моим логином и аватаркой и засчитывается в мою активность. Если адрес не привязан, коммит к аккаунту не привязывается.

**1.6. Какая команда генерирует SSH-ключ для доступа к репозиторию по SSH? (Рисунок 83)**

`ssh-keygen`, например:

```bash
ssh-keygen -t ed25519 -C "your_email@example.com"
```

Публичный ключ (`~/.ssh/id_ed25519.pub`) затем добавляется в GitHub: Settings → SSH and GPG keys → New SSH key.

## 3. Рабочий процесс с использованием GitHub

**3. 8 шагов, чтобы внести вклад в чужой проект**

1. Сделать fork проекта.
2. Создать тематическую ветку от основной ветки (`main` / `master`).
3. Сделать коммиты с улучшениями.
4. Отправить ветку в свой форк на GitHub (`git push`).
5. Открыть pull request в исходный проект.
6. Обсудить изменения и при необходимости добавить новые коммиты.
7. Владелец проекта сливает (merge) или закрывает pull request.
8. Синхронизировать свой форк с обновлённой основной веткой исходного проекта.

**3.1. Какие практики принято соблюдать, чтобы issue закрывался автоматически?**

- **При создании pull request** — в описании PR написать ключевое слово и номер или ссылку на issue: `Closes #123`, `Fixes #123` или `Resolves #123`. Issue закроется, когда PR сольют в основную ветку.
- **При создании коммита** — то же ключевое слово указать в описании (теле) коммита, а не в заголовке: `Closes #123`. Issue закроется, когда коммит попадёт в основную ветку.

Заголовок коммита и PR пишут по Conventional Commits без номера issue. Если работа сделана частично, пишут `Refs #123`: ссылка появится, но issue останется открытым.

**3.2. Как отклонить или закрыть pull request?**

Открыть pull request на вкладке **Conversation**, прокрутить вниз до поля комментария и нажать кнопку **Close pull request**. Можно сначала написать комментарий с причиной: тогда кнопка называется **Close with comment**. Закрытый PR можно снова открыть кнопкой **Reopen pull request**.



**3.3. Нужно ли создавать issue перед отправкой pull request?**

Не обязательно: pull request можно открыть и без issue, по сути это issue с прикреплённым кодом. Но для заметных изменений лучше сначала создать issue и обсудить идею с мейнтейнерами, чтобы не делать работу, которую не примут. Многие проекты требуют это в CONTRIBUTING, и тогда PR привязывают к issue ключевым словом.

**3.4. На какой вкладке можно посмотреть список изменений в pull request? (Рисунок 92)**

На вкладке **Files changed**.

**3.5. На какой вкладке находится страница обсуждения pull request? (Рисунок 94)**

На вкладке **Conversation**.

## 4. Создание pull request

**4. Можно ли открыть pull request, если вы ничего не изменили в форке?**

Нет. Pull request показывает разницу между ветками, и если её нет, GitHub пишет, что сравнивать нечего, и кнопки создания PR не будет. Нужен хотя бы один коммит, которого нет в целевой ветке.

**4.1. Что нужно сделать, чтобы открыть pull request? (Рисунок 90)**

Отправить ветку с коммитами в свой форк и открыть его на GitHub. Появится подсказка о недавно отправленной ветке с кнопкой **Compare & pull request**; другой путь — вкладка **Pull requests** → **New pull request**. Затем выбрать base (куда) и compare (откуда), заполнить заголовок и описание и нажать **Create pull request**.

**4.2. Что делать, если ваш форк устарел?**

Обновить его из исходного репозитория: нажать **Sync fork** → **Update branch** на странице форка или сделать это локально через remote `upstream` (см. 6.1). Если это мешает открытому PR, нужно влить свежую основную ветку оригинала в свою ветку и отправить её: PR обновится сам.

**4.3. Что делать, если в pull request есть конфликты слияния? (Рисунок 96)**

Если GitHub пишет, что PR нельзя слить автоматически, конфликты решают в своей ветке:

```bash
git remote add upstream https://github.com/<владелец>/<репозиторий>.git
git fetch upstream
git merge upstream/main
# исправить конфликтующие файлы
git add .
git commit
git push origin <ветка>
```

Вместо `merge` можно сделать `rebase` на свежую основную ветку. После `push` PR обновится и станет доступен для слияния. Простые конфликты можно решить и в браузере кнопкой **Resolve conflicts**.

## 6. Поддержание репозитория в актуальном состоянии

**6. Как понять, что ваш форк устарел?**

Форк не обновляется сам. На главной странице форка GitHub показывает строку вида **This branch is N commits behind <владелец>:main**: значит, в исходном репозитории есть коммиты, которых нет в форке. Локально это видно через `git fetch upstream` и `git log main..upstream/main`.

**6.1. Как обновить форк?**

В браузере: на странице форка нажать **Sync fork** → **Update branch**.

Локально:

```bash
git remote add upstream https://github.com/<владелец>/<репозиторий>.git
git fetch upstream
git switch main
git merge upstream/main
git push origin main
```

Чтобы не делать это каждый раз вручную, можно настроить `main` на отслеживание `upstream/main`: `git branch --set-upstream-to=upstream/main main`.

## 9. Особые файлы

**9. Что такое файл README и зачем он нужен?**

README (`README.md`, `README.rst`, `README`) — главный файл описания проекта, GitHub показывает его на главной странице репозитория. В нём обычно пишут, для чего проект, как его установить и использовать, какие у него зависимости, лицензия и как связаться с авторами или помочь проекту.

**9.1. Что такое файл CONTRIBUTING и зачем он нужен? (Рисунок 122)**

`CONTRIBUTING` (`CONTRIBUTING.md`) — правила для тех, кто хочет внести вклад: как оформлять issue и pull request, стиль кода и коммитов, как запускать тесты, нужно ли подписывать CLA. При создании PR или issue GitHub показывает ссылку на этот файл, чтобы участник прочитал правила до отправки.

## 10. Управление проектом

**10. Как изменить основную ветку? (Рисунок 123)**

**Settings** → **General** → раздел **Default branch** → кнопка переключения ветки (⇄) → выбрать ветку → **Update** и подтвердить. Новая ветка станет базовой для новых pull request и будет открываться по умолчанию при просмотре и клонировании репозитория.

**10.1. Как передать проект другому владельцу? Какая кнопка за это отвечает? (Рисунок 124)**

**Settings** → **General** → внизу раздел **Danger Zone** → **Transfer ownership**, кнопка **Transfer**. Нужно указать нового владельца (пользователя или организацию) и подтвердить именем репозитория. GitHub перенесёт репозиторий со всеми issues, PR и звёздами и настроит перенаправление со старого адреса.

**10.2. Что такое файл `.gitignore`?**

`.gitignore` — файл с шаблонами путей, которые Git не должен отслеживать: виртуальные окружения (`.venv/`), кэш (`__pycache__/`), файлы с секретами (`.env`), логи и результаты сборки. Шаблоны действуют только на ещё не отслеживаемые файлы: уже закоммиченный файл нужно сначала убрать из индекса командой `git rm --cached`.
