1.1. Что такое GitHub?

GitHub — это онлайн-платформа для хранения Git-репозиториев, совместной разработки программного обеспечения, отслеживания ошибок и управления проектами.

1.2. Как GitHub связан с Git?

Git — это распределённая система контроля версий, а GitHub — платформа, использующая Git для хранения репозиториев и организации совместной работы над кодом.

1.3. Чем fork репозитория отличается от его клонирования (clone)?

Fork создаёт копию репозитория в нашем аккаунте GitHub, а clone скачивает репозиторий на локальный компьютер. Fork позволяет независимо развивать проект на GitHub, а clone — работать с ним локально.

1.4. Зачем нужны pull request и как они работают?

Pull request нужен для предложения изменений в репозиторий. Разработчик создаёт отдельную ветку, вносит изменения и открывает pull request. После проверки и обсуждения изменений владелец репозитория может объединить их с основной веткой.

1.5. Использует ли GitHub ваш адрес электронной почты для привязки Git-коммитов к вашей учётной записи?

Да. GitHub сопоставляет адрес электронной почты автора коммита с адресами, привязанными к учётной записи. Можно также использовать специальный адрес noreply, чтобы не раскрывать личную почту.

3.1. Какие практики принято соблюдать, чтобы issue закрывался автоматически?

При создании pull request в его описании указывают ключевое слово и номер issue, например Closes #10 или Fixes #10.
При создании коммита аналогичную запись добавляют в сообщение коммита, например Closes #10.
Issue автоматически закрывается после попадания соответствующих изменений в основную ветку репозитория. Для автоматической привязки через описание PR целевой веткой должна быть ветка по умолчанию. 

3.2. Как отклонить или закрыть pull request? Приложите скриншот того, где это находится на GitHub.

Открыть репозиторий, перейти во вкладку Pull requests, выбрать нужный PR и на вкладке Conversation в нижней части страницы нажать кнопку Close pull request. Изменения не будут объединены с основной веткой.

![alt text](images/image.png)

3.3. Нужно ли создавать issue перед отправкой pull request?

Нет, создание issue не является обязательным. Можно сразу открыть pull request. Однако для существенных изменений рекомендуется сначала создать или обсудить issue, чтобы согласовать задачу с сопровождающими проекта.

3.4. На какой вкладке можно посмотреть список изменений в pull request?

На вкладке Files changed.

3.5. На какой вкладке находится страница обсуждения pull request?

На вкладке Conversation.

4. Можно ли открыть pull request, если вы ничего не изменили в форке?

Нет, если между сравниваемыми ветками нет различий, GitHub не позволит создать обычный pull request, поскольку нечего объединять.

4.1. Что нужно сделать, чтобы открыть pull request?

Отправить изменения в свою ветку на GitHub, перейти в Pull requests → New pull request, выбрать исходную и целевую ветки, нажать Create pull request, указать заголовок и описание, затем подтвердить создание.

4.2. Что делать, если ваш форк устарел?

Синхронизировать его с исходным репозиторием. Для этого открыть fork на GitHub, нажать Sync fork → Update branch. Также можно получить изменения из удалённого репозитория upstream с помощью Git. 

4.3. Что делать, если в pull request есть конфликты слияния?

Необходимо разрешить конфликты вручную или через кнопку Resolve conflicts, если она доступна. Затем сохранить исправления, создать коммит и обновить ветку PR. После разрешения всех конфликтов слияние станет возможным, если остальные проверки пройдены.

5.1. Какую клавишу нажать, чтобы выделенный текст вставился в комментарий как цитата?

5.2. Как вставить картинку в issue?

Перетащить изображение в поле комментария (drag and drop), вставить его из буфера обмена (Ctrl + V) или прикрепить через кнопку Attach files. GitHub автоматически загрузит изображение и добавит ссылку на него. 

6. Как понять, что ваш форк устарел?

GitHub показывает, что ветка форка отстаёт от исходного репозитория, например This branch is 3 commits behind. Это означает, что в исходном репозитории появились коммиты, отсутствующие в форке.

6.1. Как обновить форк?

Открыть свой fork и нажать Sync fork → Update branch. GitHub синхронизирует выбранную ветку с исходным репозиторием, если обновление не вызывает конфликтов.

7. Как добавить участников в репозиторий, чтобы команда могла работать над ним вместе?

Открыть репозиторий → Settings → Collaborators (или Collaborators & teams) → Add people. Затем указать имя пользователя GitHub и отправить приглашение. После принятия приглашения пользователь получит доступ к репозиторию в соответствии с назначенными правами. 

8. Какой символ используется для упоминания пользователя?

9. Что такое файл README и зачем он нужен?

README.md — файл с основной информацией о проекте. Обычно содержит описание проекта, инструкции по установке и запуску, примеры использования и другую полезную документацию. GitHub автоматически отображает его на главной странице репозитория.

9.1. Что такое файл CONTRIBUTING и зачем он нужен?

CONTRIBUTING.md — файл с правилами участия в разработке проекта. В нём описывают порядок создания issues и pull requests, требования к оформлению кода, тестированию и сообщениям коммитов. Он помогает участникам соблюдать единые стандарты разработки.

10. Как изменить основную ветку?

Открыть репозиторий → Settings → Default branch, выбрать другую ветку и подтвердить изменение. Для этого необходимы права администратора репозитория. 

10.1. Как передать проект другому владельцу? Какая кнопка за это отвечает?

Открыть репозиторий → Settings → General → Danger Zone → Transfer. Указать нового владельца и подтвердить передачу репозитория. Для передачи нужны административные права. 

10.2. Что такое файл .gitignore?

.gitignore — файл, содержащий правила исключения файлов и папок из отслеживания Git. Например, в него добавляют временные файлы, кеш, виртуальные окружения и локальные настройки.

Файл .gitignore не удаляет файлы из репозитория и не прекращает отслеживание файлов, уже добавленных в Git.
