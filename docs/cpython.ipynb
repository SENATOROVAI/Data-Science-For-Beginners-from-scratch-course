11. Какая команда используется для запуска файла с помощью CPython?

12. Можно ли запускать текстовые файлы через интерпретатор Python? Почему?

Да, если внутри валидный код:

13. Как указать путь к интерпретатору и к файлу для выполнения кода?

Указать оба пути через пробел в терминале: /путь/python /путь/файл.py.

14. Чем PyPy отличается от CPython?

PyPy быстрее за счёт JIT-компилятора, но хуже дружит с C-библиотеками.

15. Почему PyPy подходит не для всех проектов на Python?

Из-за плохой совместимости с C-библиотеками, расхода RAM и долгого старта.

16. Где можно скачать PyPy?

На официальном сайте pypy.org, а также через Conda, pyenv, uv или Docker.

17. Как установить PyPy после скачивания?

Распаковать архив, добавить папку в PATH и активировать pip.

18. Как запустить файл с помощью PyPy?

В терминале выполните команду pypy3 имя_файла.py или pypy файл.py.

19. Почему PyPy выполняет код быстрее, чем CPython?

Благодаря JIT-компилятору, переводящему горячий код в машинный.
