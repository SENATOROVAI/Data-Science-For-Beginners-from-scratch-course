1. Зачем нужна кнопка History и какие функции появляются при нажатии правой кнопкой мыши на коммит?
Нужна для просмотра истории изменений репозитория.
Команды:
Create Branch from Commit — создать новую ветку на основе выбранного коммита.
Checkout Commit — перейти к состоянию репозитория на момент выбранного коммита.
Revert Changes in Commit — отменить изменения, внесённые этим коммитом.
Cherry-pick — перенести изменения выбранного коммита в текущую ветку.
Copy SHA — скопировать уникальный идентификатор коммита.
