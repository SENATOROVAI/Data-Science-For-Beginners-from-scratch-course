Содержит вопросы и ответы на них.
1. CPython — это официальная (эталонная) реализация Python, написанная на C. 
Python — сам язык программирования и его спецификация. CPython — конкретная 
программа-интерпретатор, которая читает код Python, компилирует его в байт-код 
и выполняет.

2. Ключевые различия по таблице:

Python — язык высокого уровня, не привязан к конкретной реализации; 
описывает конструкции for, if, функции.

CPython — официальная реализация Python на C; исполняет байт-код Python; 
интерпретирует и выполняет конструкции языка.

3. Реализаций Python существует много (CPython, PyPy, Jython, IronPython, 
MicroPython, GraalPy и др.), точное число постоянно меняется. Самая 
популярная и стандартная — CPython.

4. CPython написан на C. Часть стандартной библиотеки при этом реализована 
на Python.

5. Изначально Python и CPython создал Гвидо ван Россум; сейчас CPython 
развивается сообществом и Python Software Foundation.

6. Строго говоря, CPython не считается быстрым по сравнению с компилируемыми 
языками и PyPy. Его относительная производительность объясняется тем, что 
исходный код компилируется в байт-код, который исполняется оптимизированным 
циклом на C; многие встроенные функции и модули реализованы на C. Но 
интерпретация и GIL ограничивают скорость.

7. Путь зависит от компьютера. Узнать можно так:

Windows: where python или py -0p

Linux/macOS: which python3 или command -v python3

Примеры: C:\Users\<user>\AppData\Local\Programs\Python\Python312\python.exe, 
/usr/bin/python3, /usr/local/bin/python3.

8. В папке include CPython находятся C-заголовочные файлы Python C API, 
прежде всего Python.h, необходимые для сборки C-расширений и встраивания 
Python.

9. Исходный код CPython: https://github.com/python/cpython

10. CPython при выполнении кода: разбирает исходный текст в AST, компилирует 
в байт-код, затем выполняет байт-код в виртуальной машине. Управление 
памятью — подсчёт ссылок + циклический сборщик мусора; есть GIL.

11. Запуск файла:

python file.py

python3 file.py

Windows: py file.py

Можно указать полный путь к интерпретатору.

12. Да, можно запускать текстовые файлы, если в них содержится корректный 
код Python. Исходники Python — это текст; интерпретатор читает его, 
компилирует и выполняет. Произвольный текст не на Python не выполнится.

13. Синтаксис:

<путь_к_интерпретатору> <путь_к_файлу>



/usr/bin/python3 /home/user/script.py
"C:\Python312\python.exe" "C:\Users\user\script.py"


14. PyPy — альтернативная реализация Python, написанная на RPython и 
использующая JIT-компиляцию. CPython — эталонная реализация на C без JIT. 
PyPy часто быстрее на длительных pure-Python нагрузках, но имеет другую 
совместимость и потребление памяти.

15. PyPy подходит не для всех проектов, потому что:

JIT требует «разогрева» и больше памяти;

не все C-расширения и библиотеки совместимы;

не все фреймворки/версии Python поддерживаются одинаково;

для коротких скриптов выигрыша может не быть.

16. Скачать PyPy: https://www.pypy.org/download.html

17. Установка PyPy после скачивания:

Linux/macOS: распаковать архив, добавить bin в PATH или запускать по 
полному пути;

Windows: распаковать архив и запускать pypy3.exe;

также можно установить через пакетный менеджер, если он есть.

18. Запуск файла через PyPy:


pypy3 file.py
или по полному пути к pypy3.

19. PyPy быстрее CPython за счёт JIT-компиляции: он анализирует 
«горячие» участки кода и компилирует их в машинный код, оптимизируя 
выполнение. CPython же в основном интерпретирует байт-код. Но PyPy 
быстрее не всегда — зависит от задачи.
